```python
missing_counts = pd.Series(
    0,
    index=sample.columns,
    dtype="int64"
)

total_rows = 0

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=100_000,
    low_memory=False,
    encoding="cp1252",
    dtype={
        "eventID": "string",
        "case Purchasing Document": "string",
        "case Item": "string",
        "case concept:name": "string"
    }
):

    # Handle empty strings and whitespace
    for col in chunk.columns:
        if pd.api.types.is_string_dtype(chunk[col].dtype):
            chunk[col] = chunk[col].replace(
                r"^\s*$",
                pd.NA,
                regex=True
            )

    # Count missing values
    missing_counts += chunk.isna().sum()

    total_rows += len(chunk)

print("Total rows:", total_rows)
print("Encoding used: Windows-1252")
```

```python
# Case-Level Missing Values Analysis

CASE_ID = "case concept:name"

spend_columns = [
    "case Spend area text",
    "case Sub spend area text",
    "case Spend classification text"
]

case_total_parts = []
case_missing_parts = []

all_three_missing_rows = 0
same_missing_pattern = True

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=100_000,
    low_memory=False,
    encoding="cp1252",
    usecols=[CASE_ID] + spend_columns,
    dtype={CASE_ID: "string"}
):

    # Normalize text values
    for col in spend_columns:
        chunk[col] = (
            chunk[col]
            .astype("string")
            .str.strip()
            .replace("", pd.NA)
        )

    # Check missing pattern
    missing_matrix = chunk[spend_columns].isna()

    # Are all three columns missing in exactly the same rows?
    if not missing_matrix.nunique(axis=1).le(1).all():
        same_missing_pattern = False

    all_three_missing = missing_matrix.all(axis=1)

    all_three_missing_rows += all_three_missing.sum()

    # Count all events per case
    case_total_parts.append(
        chunk.groupby(CASE_ID).size()
    )

    # Count affected events per case
    case_missing_parts.append(
        chunk.loc[all_three_missing]
             .groupby(CASE_ID).size()
    )


# Combine counts across chunks
case_total = (
    pd.concat(case_total_parts)
      .groupby(level=0)
      .sum()
)

case_missing = (
    pd.concat(case_missing_parts)
      .groupby(level=0)
      .sum()
)

# Check whether every event in an affected case is missing
total_affected_case_events = case_total.reindex(
    case_missing.index
)

fully_missing_cases = (
    total_affected_case_events == case_missing
)

print("All three columns missing in the same rows:",
      same_missing_pattern)

print("Rows with all three columns missing:",
      all_three_missing_rows)

print("Affected purchase cases:",
      len(case_missing))

print("Cases with all events missing:",
      fully_missing_cases.sum())

print("Cases with partial missing values:",
      (~fully_missing_cases).sum())
```

```python
# Timestamp Range Investigation

START_DATE = pd.Timestamp(
    "2018-01-01 00:00:00",
    tz="UTC"
)

END_DATE = pd.Timestamp(
    "2019-01-27 23:59:59",
    tz="UTC"
)

case_parts = []
outlier_samples = []

required_columns = [
    "case concept:name",
    "eventID",
    "event concept:name",
    "event time:timestamp"
]

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=100_000,
    encoding="cp1252",
    low_memory=False,
    usecols=lambda c: c.strip() in required_columns,
    dtype="string"
):

    # Clean column names
    chunk.columns = chunk.columns.str.strip()

    # Parse timestamps
    chunk["parsed_timestamp"] = pd.to_datetime(
        chunk["event time:timestamp"],
        errors="coerce",
        utc=True,
        format="mixed"
    )

    # Aggregate dates for each case in this chunk
    grouped = chunk.groupby("case concept:name").agg(
        case_start=("parsed_timestamp", "min"),
        case_end=("parsed_timestamp", "max"),
        event_count=("eventID", "size")
    )

    case_parts.append(grouped)

    # Collect sample events outside the reference window
    mask = (
        (chunk["parsed_timestamp"] < START_DATE) |
        (chunk["parsed_timestamp"] > END_DATE)
    )

    outside = chunk.loc[
        mask,
        [
            "eventID",
            "case concept:name",
            "event concept:name",
            "event time:timestamp",
            "parsed_timestamp"
        ]
    ]

    if not outside.empty:
        outlier_samples.append(
            outside.nsmallest(5, "parsed_timestamp")
        )
        outlier_samples.append(
            outside.nlargest(5, "parsed_timestamp")
        )


# Combine case-level results from all chunks
case_stats = (
    pd.concat(case_parts)
      .groupby(level=0)
      .agg(
          case_start=("case_start", "min"),
          case_end=("case_end", "max"),
          event_count=("event_count", "sum")
      )
)

# Identify cases outside the reference period
out_of_range = (
    (case_stats["case_start"] < START_DATE) |
    (case_stats["case_end"] > END_DATE)
)

# Summary
print("Total cases:", len(case_stats))

print("Cases outside reference period:",
      out_of_range.sum())

print("Cases inside reference period:",
      (~out_of_range).sum())

print(
    "Events in outside cases:",
    case_stats.loc[out_of_range, "event_count"].sum()
)

print(
    "Events in inside cases:",
    case_stats.loc[~out_of_range, "event_count"].sum()
)
```

```python
# Activity Analysis on the validated dataset

analysis_case_ids = set(
    case_stats.index[~out_of_range].astype(str)
)

activity_parts = []

total_valid_events = 0

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=100_000,
    encoding="cp1252",
    low_memory=False,
    usecols=lambda c: c.strip() in [
        "case concept:name",
        "event concept:name",
        "event User",
        "event org:resource"
    ],
    dtype="string"
):

    # Normalize column names
    chunk.columns = chunk.columns.str.strip()

    # Retain only cases in our analytical scope
    chunk = chunk[
        chunk["case concept:name"].isin(
            analysis_case_ids
        )
    ]

    total_valid_events += len(chunk)

    # Classify user recording status
    user = chunk["event User"].str.strip()

    chunk["user_not_recorded"] = (
        user.isna()
        | user.str.upper().eq("NONE").fillna(False)
    )

    # Aggregate by activity
    part = chunk.groupby(
        "event concept:name"
    ).agg(
        event_count=("event concept:name", "size"),
        user_not_recorded_count=(
            "user_not_recorded", "sum"
        )
    )

    activity_parts.append(part)


# Combine partial results
activity_profile = (
    pd.concat(activity_parts)
    .groupby(level=0)
    .sum()
)

# Calculate percentages
activity_profile["event_percentage"] = (
    activity_profile["event_count"]
    / total_valid_events * 100
).round(2)

activity_profile["user_not_recorded_percentage"] = (
    activity_profile["user_not_recorded_count"]
    / activity_profile["event_count"] * 100
).round(2)

# Sort
activity_profile = activity_profile.sort_values(
    "event_count",
    ascending=False
)

print("Total valid events:", total_valid_events)
print("Unique activities:", len(activity_profile))

display(activity_profile)
```

```python
# User and Resource Analysis

target_activities = [
    "Vendor creates invoice",
    "Record Service Entry Sheet",
    "Vendor creates debit memo",
    "Clear Invoice"
]

actor_parts = []
resource_parts = []

required_columns = [
    "case concept:name",
    "event concept:name",
    "event User",
    "event org:resource"
]

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=100_000,
    encoding="cp1252",
    low_memory=False,
    usecols=lambda c: c.strip() in required_columns,
    dtype="string"
):

    chunk.columns = chunk.columns.str.strip()

    # Keep validated cases only
    chunk = chunk[
        chunk["case concept:name"].isin(
            analysis_case_ids
        )
    ]

    # Focus on selected activities
    chunk = chunk[
        chunk["event concept:name"].isin(
            target_activities
        )
    ].copy()

    # Normalize User and Resource values
    user = (
        chunk["event User"]
        .str.strip()
    )

    resource = (
        chunk["event org:resource"]
        .str.strip()
    )

    # Treat NONE as not recorded
    user_missing = (
        user.isna()
        | user.str.upper().eq("NONE").fillna(False)
    )

    resource_missing = (
        resource.isna()
        | resource.str.upper().eq("NONE").fillna(False)
    )

    chunk["User_Status"] = np.where(
        user_missing,
        "Not Recorded",
        "Recorded"
    )

    chunk["Resource_Status"] = np.where(
        resource_missing,
        "Not Recorded",
        "Recorded"
    )

    chunk["Resource_Value"] = resource

    # Aggregate event-level user/resource status
    actor_summary = (
        chunk.groupby("event concept:name")
        .agg(
            Event_Count=("event concept:name", "size"),
            User_Not_Recorded=(
                "User_Status",
                lambda x: (x == "Not Recorded").sum()
            ),
            Resource_Not_Recorded=(
                "Resource_Status",
                lambda x: (x == "Not Recorded").sum()
            )
        )
    )

    actor_parts.append(actor_summary)

    # Resource distribution
    resource_summary = (
        chunk.loc[
            ~resource_missing,
            ["event concept:name", "Resource_Value"]
        ]
        .groupby(
            ["event concept:name", "Resource_Value"]
        )
        .size()
        .rename("Event_Count")
        .reset_index()
    )

    resource_parts.append(resource_summary)


# Combine results
actor_profile = (
    pd.concat(actor_parts)
    .groupby(level=0)
    .sum()
)

resource_profile = (
    pd.concat(resource_parts)
    .groupby(
        ["event concept:name", "Resource_Value"]
    )["Event_Count"]
    .sum()
    .reset_index()
)

display(actor_profile)

display(
    resource_profile.sort_values(
        ["event concept:name", "Event_Count"],
        ascending=[True, False]
    )
)
```

```python
# Process Variant Preparation

required_columns = [
    "case concept:name",
    "event concept:name",
    "event time:timestamp"
]

event_parts = []
source_row = 0

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=100_000,
    encoding="cp1252",
    low_memory=False,
    dtype="string",
    usecols=lambda c: c.strip() in required_columns
):

    chunk.columns = chunk.columns.str.strip()

    # Keep only validated cases
    chunk = chunk[
        chunk["case concept:name"].isin(
            analysis_case_ids
        )
    ].copy()

    # Preserve original CSV row order
    chunk["source_row"] = np.arange(
        source_row,
        source_row + len(chunk)
    )

    source_row += 100_000

    # Correct date interpretation
    chunk["timestamp"] = pd.to_datetime(
        chunk["event time:timestamp"],
        format="mixed",
        dayfirst=True,
        errors="coerce"
    )

    # Retain usable process events
    chunk = chunk.dropna(
        subset=[
            "case concept:name",
            "event concept:name",
            "timestamp"
        ]
    )

    event_parts.append(
        chunk[
            [
                "case concept:name",
                "event concept:name",
                "timestamp",
                "source_row"
            ]
        ]
    )


process_events = pd.concat(
    event_parts,
    ignore_index=True
)

print("Process events:", len(process_events))

print(
    "Unique cases:",
    process_events["case concept:name"].nunique()
)

display(process_events.head())
```

```python
timestamp_counts = (
    process_events
    .groupby(
        ["case concept:name", "timestamp"]
    )
    .size()
)

tied_groups = timestamp_counts[
    timestamp_counts > 1
]

print(
    "Groups with same timestamp:",
    len(tied_groups)
)

print(
    "Cases with tied timestamps:",
    tied_groups.index
    .get_level_values(0)
    .nunique()
)

print(
    "Events in tied groups:",
    tied_groups.sum()
)
```

```python
from collections import Counter

# Build activity sequence for every purchase case
case_variants = (
    process_events
    .groupby("case concept:name", sort=False)["event concept:name"]
    .agg(lambda x: tuple(x.astype(str).tolist()))
)

# Count process variants using Python Counter
variant_counter = Counter(
    tuple(sequence)
    for sequence in case_variants.tolist()
)

print("Total cases:", len(case_variants))
print("Unique process variants:", len(variant_counter))
```

```python
top_variants = variant_counter.most_common(15)

variant_rows = []

for rank, (sequence, count) in enumerate(
    top_variants,
    start=1
):
    variant_rows.append({
        "Rank": rank,
        "Activity_Sequence": " -> ".join(sequence),
        "Number_of_Cases": count,
        "Percentage": round(
            count / len(case_variants) * 100,
            2
        )
    })

variant_summary = pd.DataFrame(variant_rows)

display(variant_summary)
```

```python
# Select the activities needed for temporal analysis

activities_to_check = [
    "Vendor creates invoice",
    "Record Goods Receipt",
    "Record Invoice Receipt",
    "Clear Invoice",
    "Remove Payment Block"
]

selected_events = process_events[
    process_events["event concept:name"].isin(activities_to_check)
].copy()

# Get the first timestamp of each activity in each case

first_event_times = (
    selected_events
    .groupby(
        ["case concept:name", "event concept:name"]
    )["timestamp"]
    .min()
    .unstack()
)

# Align with all cases in our analysis

first_event_times = first_event_times.reindex(features.index)

# Define the activity pairs to compare

pairs = [
    ("Vendor creates invoice", "Record Goods Receipt"),
    ("Record Goods Receipt", "Record Invoice Receipt"),
    ("Record Invoice Receipt", "Clear Invoice"),
    ("Remove Payment Block", "Clear Invoice")
]

temporal_rows = []

for activity_a, activity_b in pairs:

    time_a = first_event_times[activity_a]
    time_b = first_event_times[activity_b]

    both_exist = time_a.notna() & time_b.notna()

    a_before_b = both_exist & (time_a < time_b)
    same_time = both_exist & (time_a == time_b)
    b_before_a = both_exist & (time_b < time_a)

    elapsed_hours = (
        time_b[a_before_b] - time_a[a_before_b]
    ).dt.total_seconds() / 3600

    temporal_rows.append({
        "Activity_A": activity_a,
        "Activity_B": activity_b,
        "Cases_With_Both": int(both_exist.sum()),
        "A_Before_B": int(a_before_b.sum()),
        "A_Before_B_Percentage": round(
            a_before_b.sum() / both_exist.sum() * 100, 2
        ) if both_exist.sum() else 0,
        "Same_Timestamp": int(same_time.sum()),
        "B_Before_A": int(b_before_a.sum()),
        "B_Before_A_Percentage": round(
            b_before_a.sum() / both_exist.sum() * 100, 2
        ) if both_exist.sum() else 0,
        "Median_Elapsed_Hours": round(
            elapsed_hours.median(), 2
        ) if not elapsed_hours.empty else None,
        "P90_Elapsed_Hours": round(
            elapsed_hours.quantile(0.90), 2
        ) if not elapsed_hours.empty else None
    })

temporal_summary = pd.DataFrame(temporal_rows)

display(temporal_summary)
```

```python
# Calculate elapsed time distributions in days

duration_pairs = {
    "Vendor Invoice to Goods Receipt": (
        "Vendor creates invoice",
        "Record Goods Receipt"
    ),
    "Goods Receipt to Invoice Receipt": (
        "Record Goods Receipt",
        "Record Invoice Receipt"
    ),
    "Invoice Receipt to Invoice Clearing": (
        "Record Invoice Receipt",
        "Clear Invoice"
    ),
    "Payment Block Removal to Invoice Clearing": (
        "Remove Payment Block",
        "Clear Invoice"
    )
}

duration_series = {}
delay_rows = []

for process_name, (start_activity, end_activity) in duration_pairs.items():

    start_time = first_event_times[start_activity]
    end_time = first_event_times[end_activity]

    valid = (
        start_time.notna()
        & end_time.notna()
        & (start_time < end_time)
    )

    days = (
        end_time[valid] - start_time[valid]
    ).dt.total_seconds() / 86400

    duration_series[process_name] = days

    delay_rows.append({
        "Process": process_name,
        "Cases": len(days),
        "Median_Days": round(days.median(), 2),
        "P90_Days": round(days.quantile(0.90), 2),
        "P95_Days": round(days.quantile(0.95), 2),
        "Over_30_Days": int((days > 30).sum()),
        "Over_30_Days_Pct": round((days > 30).mean() * 100, 2),
        "Over_60_Days": int((days > 60).sum()),
        "Over_60_Days_Pct": round((days > 60).mean() * 100, 2),
        "Over_90_Days": int((days > 90).sum()),
        "Over_90_Days_Pct": round((days > 90).mean() * 100, 2)
    })

delay_analysis = pd.DataFrame(delay_rows)

display(delay_analysis)
```

```python
# 1. Build an invoice-level case dataset

receipt_time = first_event_times["Record Invoice Receipt"]
clearing_time = first_event_times["Clear Invoice"]

both_events_exist = (
    receipt_time.notna() & clearing_time.notna()
)

clearing_after_receipt = (
    both_events_exist & (clearing_time > receipt_time)
)

invoice_cycle_days = (
    (clearing_time - receipt_time)
    .dt.total_seconds() / 86400
).where(clearing_after_receipt)

invoice_cases = pd.DataFrame(index=first_event_times.index)

invoice_cases["Invoice_Receipt_Recorded"] = receipt_time.notna()

invoice_cases["Invoice_Clearing_Recorded"] = clearing_time.notna()

invoice_cases["Clearing_After_Receipt"] = clearing_after_receipt

invoice_cases["Invoice_Cycle_Days"] = invoice_cycle_days


# 2. Join validated case metadata

metadata_for_join = case_metadata.copy()

invoice_cases.index = invoice_cases.index.astype(str)
metadata_for_join.index = metadata_for_join.index.astype(str)

invoice_cases = invoice_cases.join(metadata_for_join)

# Keep only cases where invoice receipt was recorded
invoice_cohort = invoice_cases[
    invoice_cases["Invoice_Receipt_Recorded"]
].copy()


# 3. Verify the analytical population

print("Cases with invoice receipt:",
      len(invoice_cohort))

print("Cases without recorded clearing:",
      int((~invoice_cohort["Invoice_Clearing_Recorded"]).sum()))

print("Cases with clearing after receipt:",
      int(invoice_cohort["Clearing_After_Receipt"].sum()))

print("Cases with both events but not strictly ordered:",
      int(
          (
              invoice_cohort["Invoice_Clearing_Recorded"]
              & ~invoice_cohort["Clearing_After_Receipt"]
          ).sum()
      ))


# 4. Define the business dimensions

dimensions = [
    "case Company",
    "case Document Type",
    "case Spend area text",
    "case Sub spend area text",
    "case Purch. Doc. Category name",
    "case Spend classification text"
]


# 5. Analyze each dimension

def summarize_invoice_groups(data, dimension):

    summary = (
        data.groupby(dimension, dropna=False)
        .agg(
            Invoice_Receipt_Cases=(
                "Invoice_Receipt_Recorded", "size"
            ),

            Cases_Without_Recorded_Clearing=(
                "Invoice_Clearing_Recorded",
                lambda s: (~s).sum()
            ),

            No_Recorded_Clearing_Pct=(
                "Invoice_Clearing_Recorded",
                lambda s: (~s).mean() * 100
            ),

            Cases_Cleared_After_Receipt=(
                "Clearing_After_Receipt", "sum"
            ),

            Median_Days=(
                "Invoice_Cycle_Days", "median"
            ),

            P90_Days=(
                "Invoice_Cycle_Days",
                lambda s: s.quantile(0.90)
            ),

            Cases_Over_30_Days=(
                "Invoice_Cycle_Days",
                lambda s: (s > 30).sum()
            ),

            Over_30_Days_Pct=(
                "Invoice_Cycle_Days",
                lambda s: (s.dropna() > 30).mean() * 100
                if s.notna().any() else float("nan")
            ),

            Over_60_Days_Pct=(
                "Invoice_Cycle_Days",
                lambda s: (s.dropna() > 60).mean() * 100
                if s.notna().any() else float("nan")
            )
        )
    )

    # Exclude small groups to reduce unstable comparisons
    eligible = summary[
        summary["Invoice_Receipt_Cases"] >= 500
    ].copy()

    return eligible.sort_values(
        "P90_Days",
        ascending=False
    )


# 6. Display the 15 groups with the highest P90 delay

for dimension in dimensions:

    print("\n" + "=" * 90)
    print("DIMENSION:", dimension)

    result = summarize_invoice_groups(
        invoice_cohort,
        dimension
    )

    print("Groups with at least 500 invoice-receipt cases:",
          len(result))

    display(result.head(15).round(2))
```

```python

import pandas as pd
from pathlib import Path

# --------------------------------------------------
# 1. Configuration
# --------------------------------------------------

CHUNK_SIZE = 100_000

CASE_COL = "case concept:name"
TIMESTAMP_COL = "event time:timestamp"

ANALYSIS_START = pd.Timestamp("2018-01-01")
ANALYSIS_END_EXCLUSIVE = pd.Timestamp("2019-01-28")

# Expected baseline from our previous analysis
EXPECTED = {
    "Raw_Events": 1_595_923,
    "Total_Cases": 251_734,
    "Excluded_Cases": 271,
    "Included_Cases": 251_463,
    "Excluded_Events": 8_121,
    "Included_Events": 1_587_802
}


# --------------------------------------------------
# 2. Extract case IDs and timestamps in chunks
# --------------------------------------------------

case_stats_parts = []

total_rows = 0
missing_case_id_rows = 0
invalid_timestamp_rows = 0

reader = pd.read_csv(
    DATA_PATH,
    encoding="cp1252",
    usecols=[CASE_COL, TIMESTAMP_COL],
    dtype={
        CASE_COL: "string",
        TIMESTAMP_COL: "string"
    },
    chunksize=CHUNK_SIZE
)

for chunk_number, chunk in enumerate(reader, start=1):

    total_rows += len(chunk)

    # Normalize case IDs
    chunk[CASE_COL] = chunk[CASE_COL].str.strip()

    missing_case_id_rows += chunk[CASE_COL].isna().sum()

    # Parse dates using the correct day-first format
    chunk[TIMESTAMP_COL] = pd.to_datetime(
        chunk[TIMESTAMP_COL],
        format="mixed",
        dayfirst=True,
        errors="coerce"
    )

    invalid_timestamp_rows += (
        chunk[TIMESTAMP_COL].isna().sum()
    )

    # Aggregate timestamps and event counts per case
    chunk_stats = (
        chunk.dropna(subset=[CASE_COL])
        .groupby(CASE_COL, sort=False)[TIMESTAMP_COL]
        .agg(
            First_Event="min",
            Last_Event="max",
            Event_Count="size"
        )
    )

    case_stats_parts.append(chunk_stats)

    print(
        f"Processed chunk {chunk_number} | "
        f"Rows processed: {total_rows:,}"
    )


# --------------------------------------------------
# 3. Data quality gate
# --------------------------------------------------

if missing_case_id_rows > 0:
    raise ValueError(
        f"Missing case IDs detected: {missing_case_id_rows}"
    )

if invalid_timestamp_rows > 0:
    raise ValueError(
        f"Invalid or missing timestamps detected: "
        f"{invalid_timestamp_rows}"
    )


# --------------------------------------------------
# 4. Combine chunk-level results
# --------------------------------------------------

case_stats = (
    pd.concat(case_stats_parts)
    .groupby(level=0, sort=False)
    .agg({
        "First_Event": "min",
        "Last_Event": "max",
        "Event_Count": "sum"
    })
)

case_stats.index.name = CASE_COL


# --------------------------------------------------
# 5. Define case-level analytical scope
# --------------------------------------------------

out_of_range = (
    (case_stats["First_Event"] < ANALYSIS_START)
    | (case_stats["Last_Event"] >= ANALYSIS_END_EXCLUSIVE)
    | case_stats["First_Event"].isna()
    | case_stats["Last_Event"].isna()
)

case_stats["In_Analysis_Scope"] = ~out_of_range

analysis_case_ids = set(
    case_stats.index[~out_of_range].astype(str)
)

excluded_case_ids = set(
    case_stats.index[out_of_range].astype(str)
)


# --------------------------------------------------
# 6. Reconcile actual results with expected baseline
# --------------------------------------------------

actual = {
    "Raw_Events": total_rows,
    "Total_Cases": len(case_stats),
    "Excluded_Cases": int(out_of_range.sum()),
    "Included_Cases": int((~out_of_range).sum()),
    "Excluded_Events": int(
        case_stats.loc[out_of_range, "Event_Count"].sum()
    ),
    "Included_Events": int(
        case_stats.loc[~out_of_range, "Event_Count"].sum()
    )
}

reconciliation_rows = []

for metric, expected_value in EXPECTED.items():

    actual_value = actual[metric]

    reconciliation_rows.append({
        "Metric": metric,
        "Expected": expected_value,
        "Actual": actual_value,
        "Status": (
            "PASS"
            if actual_value == expected_value
            else "REVIEW"
        )
    })

scope_validation = pd.DataFrame(reconciliation_rows)


# --------------------------------------------------
# 7. Save reproducible intermediate outputs
# --------------------------------------------------

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
QUALITY_DIR = PROJECT_ROOT / "data" / "quality"

case_stats.to_csv(
    PROCESSED_DIR / "case_scope.csv",
    index=True
)

scope_validation.to_csv(
    QUALITY_DIR / "scope_validation.csv",
    index=False
)


# --------------------------------------------------
# 8. Final report
# --------------------------------------------------

print("\n" + "=" * 60)
print("EXTRACTION AND SCOPE VALIDATION")
print("=" * 60)

print("Missing case IDs:", missing_case_id_rows)
print("Invalid timestamps:", invalid_timestamp_rows)

display(scope_validation)

print("\nAnalysis case IDs:", len(analysis_case_ids))
print("Excluded case IDs:", len(excluded_case_ids))

print("\nSaved outputs:")
print(PROCESSED_DIR / "case_scope.csv")
print(QUALITY_DIR / "scope_validation.csv")
```

```python

from pathlib import Path
import pandas as pd
import numpy as np

# --------------------------------------------------
# 1. Confirm that we are using existing analysis
# --------------------------------------------------

required_objects = [
    "process_events",
    "case_metadata",
    "case_stats",
    "invoice_cohort",
    "first_event_times",
    "features",
    "analysis_case_ids"
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "These existing analysis objects are missing: "
        + ", ".join(missing_objects)
        + ". Do not rerun the full analysis yet."
    )

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
QUALITY_DIR = PROJECT_ROOT / "data" / "quality"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
QUALITY_DIR.mkdir(parents=True, exist_ok=True)

# --------------------------------------------------
# 2. Build the event fact table from process_events
# --------------------------------------------------

events = process_events.rename(columns={
    "case concept:name": "case_id",
    "event concept:name": "activity",
    "timestamp": "event_timestamp"
}).copy()

events = events[
    ["source_row", "case_id", "activity", "event_timestamp"]
].copy()

events["source_row"] = events["source_row"].astype("int64")
events["case_id"] = events["case_id"].astype("string")
events["activity"] = events["activity"].astype("string")

# --------------------------------------------------
# 3. Build the procurement case dimension
# --------------------------------------------------

scope_ids = set(map(str, analysis_case_ids))

case_dim = case_metadata.copy()
case_dim.index = case_dim.index.astype(str)

case_stats_clean = case_stats.copy()
case_stats_clean.index = case_stats_clean.index.astype(str)

case_dim = case_dim.loc[
    case_dim.index.isin(scope_ids)
].join(
    case_stats_clean.loc[
        case_stats_clean.index.isin(scope_ids),
        ["First_Event", "Last_Event", "Event_Count"]
    ],
    how="inner"
)

case_dim.index.name = "case_id"

case_dim = case_dim.rename(columns={
    "case Company": "company",
    "case Document Type": "document_type",
    "case Spend area text": "spend_area",
    "case Sub spend area text": "sub_spend_area",
    "case Spend classification text": "spend_classification",
    "case Purch. Doc. Category name": "purchase_document_category",
    "First_Event": "first_event_at",
    "Last_Event": "last_event_at",
    "Event_Count": "event_count"
}).reset_index()

case_dim["event_count"] = case_dim["event_count"].astype("int64")

# --------------------------------------------------
# 4. Build invoice monitoring mart
# --------------------------------------------------

invoice_source = invoice_cohort.copy()
invoice_source.index = invoice_source.index.astype(str)

event_times = first_event_times.copy()
event_times.index = event_times.index.astype(str)

features_clean = features.copy()
features_clean.index = features_clean.index.astype(str)

invoice = pd.DataFrame(index=invoice_source.index)

# Business dimensions
dimension_mapping = {
    "case Company": "company",
    "case Document Type": "document_type",
    "case Spend area text": "spend_area",
    "case Sub spend area text": "sub_spend_area",
    "case Spend classification text": "spend_classification"
}

for source_col, output_col in dimension_mapping.items():
    invoice[output_col] = invoice_source[source_col]

# Relevant timestamps
invoice["invoice_receipt_at"] = (
    event_times["Record Invoice Receipt"].reindex(invoice.index)
)

invoice["invoice_clearing_at"] = (
    event_times["Clear Invoice"].reindex(invoice.index)
)

invoice["payment_block_removed_at"] = (
    event_times["Remove Payment Block"].reindex(invoice.index)
)

receipt = invoice["invoice_receipt_at"]
clearing = invoice["invoice_clearing_at"]

has_clearing = clearing.notna()

cleared_after_receipt = (
    receipt.notna()
    & clearing.notna()
    & (clearing > receipt)
)

# Preserve the three different outcomes
invoice["invoice_clearing_status"] = np.select(
    [
        cleared_after_receipt,
        has_clearing & ~cleared_after_receipt,
        ~has_clearing
    ],
    [
        "CLEARED_AFTER_RECEIPT",
        "CLEARING_NOT_AFTER_RECEIPT",
        "NO_RECORDED_CLEARING"
    ],
    default="REVIEW"
)

elapsed_days = (
    (clearing - receipt).dt.total_seconds() / 86400
)

invoice["invoice_cycle_days"] = (
    elapsed_days.where(cleared_after_receipt)
)

# Activity indicators from our existing features table
feature_mapping = {
    "Goods_Receipt": "goods_receipt",
    "Vendor_Invoice": "vendor_invoice",
    "Invoice_Cleared": "invoice_cleared",
    "Payment_Block_Removed": "payment_block_removed",
    "Purchase_Requisition": "purchase_requisition",
    "Service_Entry_Sheet": "service_entry_sheet",
    "Price_Changed": "price_changed",
    "Quantity_Changed": "quantity_changed",
    "Deletion_Activity": "deletion_activity"
}

for source_col, output_col in feature_mapping.items():
    invoice[output_col] = (
        features_clean[source_col]
        .reindex(invoice.index)
        .astype("boolean")
    )

# Delay indicators remain NULL when no valid cycle time exists
for days in [30, 60, 90]:
    col = f"over_{days}_days"
    invoice[col] = (
        invoice["invoice_cycle_days"]
        .gt(days)
        .astype("boolean")
        .where(invoice["invoice_cycle_days"].notna())
    )

# Time from payment-block removal to clearing
block_time = invoice["payment_block_removed_at"]

block_to_clear = (
    (clearing - block_time).dt.total_seconds() / 86400
)

valid_block_sequence = (
    block_time.notna()
    & clearing.notna()
    & (clearing > block_time)
)

invoice["payment_block_to_clearing_days"] = (
    block_to_clear.where(valid_block_sequence)
)

invoice.index.name = "case_id"
invoice = invoice.reset_index()

# --------------------------------------------------
# 5. Save the three data products
# --------------------------------------------------

output_tables = {
    "fact_procurement_events.parquet": events,
    "dim_procurement_cases.parquet": case_dim,
    "mart_invoice_monitoring.parquet": invoice
}

for filename, dataframe in output_tables.items():

    final_path = PROCESSED_DIR / filename
    temp_path = PROCESSED_DIR / f"{filename}.tmp"

    dataframe.to_parquet(
        temp_path,
        engine="pyarrow",
        compression="snappy",
        index=False
    )

    temp_path.replace(final_path)

# --------------------------------------------------
# 6. Validate the outputs
# --------------------------------------------------

status_counts = invoice[
    "invoice_clearing_status"
].value_counts()

checks = [
    ("Event rows", len(events), 1_587_802),
    ("Unique event source rows", events["source_row"].nunique(), 1_587_802),
    ("Event cases", events["case_id"].nunique(), 251_463),
    ("Case dimension rows", len(case_dim), 251_463),
    ("Unique case IDs", case_dim["case_id"].nunique(), 251_463),
    ("Sum of case event counts", int(case_dim["event_count"].sum()), 1_587_802),
    ("Invoice mart rows", len(invoice), 211_113),
    ("Cleared after receipt", int((invoice["invoice_clearing_status"] == "CLEARED_AFTER_RECEIPT").sum()), 183_009),
    ("No recorded clearing", int((invoice["invoice_clearing_status"] == "NO_RECORDED_CLEARING").sum()), 28_081),
    ("Clearing not after receipt", int((invoice["invoice_clearing_status"] == "CLEARING_NOT_AFTER_RECEIPT").sum()), 23),
    ("Missing event timestamps", int(events["event_timestamp"].isna().sum()), 0),
    ("Missing event case IDs", int(events["case_id"].isna().sum()), 0),
    ("Events outside case dimension", int((~events["case_id"].isin(case_dim["case_id"])).sum()), 0)
]

quality_report = pd.DataFrame(
    checks,
    columns=["Check", "Actual", "Expected"]
)

quality_report["Status"] = np.where(
    quality_report["Actual"] == quality_report["Expected"],
    "PASS",
    "FAIL"
)

quality_report.to_csv(
    QUALITY_DIR / "etl_validation.csv",
    index=False
)

# --------------------------------------------------
# 7. Preserve previous analytical outputs
# --------------------------------------------------

optional_reports = {
    "top_process_variants.csv": "variant_summary",
    "activity_coverage.csv": "feature_summary",
    "activity_relationships.csv": "relationship_summary",
    "temporal_analysis.csv": "temporal_summary",
    "delay_analysis.csv": "delay_analysis"
}

for filename, variable_name in optional_reports.items():

    dataframe = globals().get(variable_name)

    if isinstance(dataframe, pd.DataFrame):
        dataframe.to_csv(
            QUALITY_DIR / filename,
            index=True
        )

print("DATA PRODUCTS CREATED")
print("-" * 55)

for filename, dataframe in output_tables.items():
    path = PROCESSED_DIR / filename
    print(
        filename,
        "| Rows:", f"{len(dataframe):,}",
        "| Size MB:", round(path.stat().st_size / (1024 ** 2), 2)
    )

print("\nETL VALIDATION:")
display(quality_report)

print(
    "All checks passed:",
    (quality_report["Status"] == "PASS").all()
)
```
